```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression # pakai logistik karena yh hasilnya survived or not survived aja
from sklearn.metrics import accuracy_score

'''Titanic Kaggle pakai logistic regression'''

#kumpulin data2nya
data_latihan = pd.read_csv('train.csv')
data_uji = pd.read_csv('test.csv')
data_latihan.head() # datanya byk yg kosong hufttt...

# cleaning data dulu, byk yg kosong
def clean_data(data):
  #sokk pilih yang akan jadi fitur (yang berpengaruh ke hasil target)
  features = ['Pclass','Sex','Age','SibSp','Parch','Fare','Embarked']
  data = data[features].copy()

  # data yang kosong diisi, ( age, fare = median , embarked = modus ajalah yaa)
  data['Age'] = data['Age'].fillna(data['Age'].median())
  data['Fare'] = data['Fare'].fillna(data['Fare'].median())
  data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])

  #ubah string ke angka (sex, embarked)
  data['Sex']= data['Sex'].map({'male':0, 'female':1})
  data['Embarked'] = data['Embarked'].map({'S': 0, 'C': 1, 'Q':2})

  return data


x = clean_data(data_latihan)
y = data_latihan['Survived'] #y nihh targettz

x_uji_asli = clean_data(data_uji)

#split dataa
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)

model = LogisticRegression(max_iter=1000)

model.fit(x_train, y_train)

#cek cek akurasi model
akurasi = model.predict(x_test)
print(f"akurasi model : {accuracy_score(y_test, akurasi)}")

prediksi = model.predict(x_uji_asli)

#hasil
hasil = pd.DataFrame({
    "PassengerId" : data_uji['PassengerId'],
    "Survived" : prediksi
})

print(hasil.head())

```
